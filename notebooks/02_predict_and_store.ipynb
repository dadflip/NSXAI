# Prédiction + stockage Fuseki (graphe recommendations)

Prérequis : notebook `01_train_recommender` exécuté ou `POST /api/recommendation/train`.

```python
import sys
from pathlib import Path
ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
sys.path.insert(0, str(ROOT))
```

```python
from nsxai.ml.predict import get_courses, predict_for_course, store_recommendations

courses = get_courses()
if not courses:
    raise RuntimeError("Aucun cours — exécutez d'abord le seed (notebook 01).")
course_uri = courses[0]["uri"]
print(f"Cours: {courses[0]['label']} → {course_uri}")
```

```python
preds = predict_for_course(course_uri, top_k=5)
stored = store_recommendations(course_uri, preds)

for p in preds:
    print(f"\n{p['gde_local']} (score={p['score']})")
    print(p['explanation']['human_readable'])
print("\nStocké:", stored)
```
