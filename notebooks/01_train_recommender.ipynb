# Entraînement ML — après configuration de la tâche

1. Exporter / peupler Fuseki (`nsxai_cli.py start all` + load)
2. Définir **ce qu'on prédit** (voir `docs/ml-task-specification.md`)
3. `prepare` → lire `prediction_summary`
4. `train`

```
import sys
from pathlib import Path
ROOT = Path.cwd().parent
sys.path.insert(0, str(ROOT))
```

```
from nsxai.ml.task import MLTaskSpec, TASK_PRESETS
from nsxai.ml.preprocess import prepare_dataset
from nsxai.ml.train import train

spec = MLTaskSpec.from_dict(TASK_PRESETS["gde_for_course"])
manifest = prepare_dataset(spec)
print("=== CE QU'ON PRÉDIT ===")
print(manifest.prediction_summary)
print(manifest.feature_summary)
print("Stats:", manifest.stats)
if manifest.warnings:
    print("Warnings:", manifest.warnings)
```

```
result = train(task=spec, min_samples=1)
print(result["prediction_summary"])
print("Modèle:", result["model_path"])
```
